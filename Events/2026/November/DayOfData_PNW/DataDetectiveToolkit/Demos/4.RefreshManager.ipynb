We will inspect the evidence, explore the controls, and verify the result of any refresh we choose to run.

```python
# Configuration: change the investigation targets here.
WORKSPACE = None  # None = the current Fabric workspace
DATASET_REFRESH = "Trudoso10K"
```

```python
import importlib.metadata
import notebookutils
import sempy.fabric as fabric
import sempy_labs as labs
import pandas as pd
from sempy_labs.semantic_model import refresh_manager

sll_version = importlib.metadata.version("semantic-link-labs")
print(f"Semantic Link Labs: {sll_version} (demo target: 0.18.1)")

context = notebookutils.runtime.context
WORKSPACE_ID = WORKSPACE or context["currentWorkspaceId"]
WORKSPACE_NAME = context["currentWorkspaceName"]
LAKEHOUSE_ID = context.get("defaultLakehouseId")
LAKEHOUSE_NAME = context.get("defaultLakehouseName")

print(f"Notebook workspace: {WORKSPACE_NAME}")
print(f"Model workspace: {WORKSPACE or WORKSPACE_NAME}")
print(f"Main model: {DATASET_REFRESH}")
```

## 2. Choose a semantic model
Start with **Trudoso10K**, or change `DATASET_REFRESH` to another model in this workspace. An Import model with working data-source credentials and existing enhanced-refresh history makes the history and execution details easier to explore.
The available refresh options depend on the model's storage mode and capacity. If history is empty, it is not evidence that the model has never refreshed.

```python
print(f"Semantic model: {DATASET_REFRESH}")
print(f"Workspace: {WORKSPACE}")
```

## 3. Open Refresh Manager
Run the next cell to open the interface. Opening the manager does not automatically start a model refresh or change its schedule.
Start by examining the recent enhanced-refresh history. Select an execution and inspect its status, duration, and available details. What evidence helps explain a slow or failed refresh?

```python
refresh_manager(
    dataset=DATASET_REFRESH,
    workspace=WORKSPACE,
    dark_mode=True,
)
```

## 4. Refresh only what needs attention
In the manager, examine the options for a full-model refresh and, where supported, selected tables or partitions.
To try a refresh, choose the intended scope and submit it through the interface. Watch the live status, then inspect the completed execution. A submitted request is not the same as a successful refresh.
Starting or canceling a refresh is a real operation on the selected model. Use a model you can refresh and a scope appropriate for its data sources.

## 5. Inspect the schedule
Open the scheduled-refresh settings. How often does this model refresh, and does the schedule match the business need?
You can inspect the settings without changing them. If you choose to edit and save a schedule, that changes the model's actual scheduled refreshes.

## 6. Confirm the outcome
If you ran a refresh, confirm that its final status is successful and review its execution details. Then query the model or refresh a connected report to check the expected data.
We now have a repeatable way to investigate refresh activity and manage refreshes from the same notebook where we inspect model quality and report dependencies.
